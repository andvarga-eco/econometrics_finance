# Modelos GARCH, entendiendo los parámetros

Sabemos que el comportamiento de la varianza condicional e incondicional de un proceso GARCH está gobernada por tres parámetros: $\omega,\quad \alpha,\quad \beta$

- $\omega$: es el componente de l+inea base de la volatilidad, es lo que evita que la volatilidad llegue a cero
- $\alpha$: mide la sensibilidad a los choques. Nos dice cuánto de un choque hoy afecta la volatilidad esperada de mañana. Valores más altos producen picos mayores de volatilidad
- $\beta$: mide la persistencia de la volatilidad. Indica que tan rápido se disipa el choque de volatilidad

A partir de estos parámetros obtenemos el pronóstico de volatilidad, así como la varianza incondicional o de largo plazo, y el tiempo que toma para que el choque descienda a la mitad de su tamaño original, half-life

- Varianza incondicional: $\sigma^2=\dfrac{\omega}{1-\alpha-\beta}$
- half-life, $h=\dfrac{ln(0.5)}{ln(\alpha+\beta)}$
- Volatilidad anual: $\sigma_{anual}=\sigma\sqrt{252}$
