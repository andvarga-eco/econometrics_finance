# Procesos simulados

En esta actividad se examinarán visualmente diferentes procesos de series de tiempo, a fin de reconocer los patrones que exhiben series estacionarias y/o con dependencia débil

## Procesos

- $e_t \quad \text{i.i.d} \sim N(0,1)$. Ruido blanco gausssiano

- $y_t=e_t+0.5e_{t-1}$

- $y_t=Z$; con $Z\sim N(0,1)$

- $y_t=\sqrt{t}e_t$

- $y_t=(-1)^tZ$

- $y_t=1+y_{t-1}+e_t$

- $y_t=1+0.8y_{t-1}+e_t$

- $y_t=y_{t-1}+e_e$

```r
## =========================================================================
## Simulación de procesos de series de tiempo
## Autor: (generado para Andrés)
## -------------------------------------------------------------------------
## En todos los casos e_t ~ N(0, sigma^2) es ruido blanco gaussiano
## =========================================================================

set.seed(123)     # reproducibilidad
n     <- 200       # tamaño de la muestra
sigma <- 1          # desviación estándar del ruido blanco

t_index <- 1:n
et      <- rnorm(n, mean = 0, sd = sigma)   # ruido blanco e_t común a los procesos

## -------------------------------------------------------------------------
## y_t = e_t + 0.5 e_{t-1} 
## -------------------------------------------------------------------------
sim_MA1 <- function(n, sigma = 1, theta = 0.5) {
  e <- rnorm(n + 1, mean = 0, sd = sigma)     # un rezago extra para e_{0}
  y <- e[2:(n + 1)] + theta * e[1:n]
  ts(y)
}
y_i <- sim_MA1(n, sigma, theta = 0.5)

## -------------------------------------------------------------------------
##  y_t = Z, con Z ~ N(0,1)
## -------------------------------------------------------------------------
sim_constZ <- function(n) {
  Z <- rnorm(1, mean = 0, sd = 1)
  ts(rep(Z, n))
}
y_ii <- sim_constZ(n)

## -------------------------------------------------------------------------
## y_t = sqrt(t) * e_t  
## -------------------------------------------------------------------------
sim_hetero <- function(n, sigma = 1) {
  e <- rnorm(n, mean = 0, sd = sigma)
  y <- sqrt(1:n) * e
  ts(y)
}
y_iv <- sim_hetero(n, sigma)

## -------------------------------------------------------------------------
## y_t = (-1)^t * Z, con Z ~ N(0,1) fijo -> oscila entre +Z y -Z
## -------------------------------------------------------------------------
sim_altern <- function(n) {
  Z <- rnorm(1, mean = 0, sd = 1)
  y <- ((-1)^(1:n)) * Z
  ts(y)
}
y_v <- sim_altern(n)

## -------------------------------------------------------------------------
##  y_t = 1 + y_{t-1} + e_t  -> caminata aleatoria CON deriva (drift = 1)
## -------------------------------------------------------------------------
sim_RWdrift <- function(n, sigma = 1, drift = 1, y0 = 0) {
  e <- rnorm(n, mean = 0, sd = sigma)
  y <- numeric(n)
  y[1] <- drift + y0 + e[1]
  for (tt in 2:n) y[tt] <- drift + y[tt - 1] + e[tt]
  ts(y)
}
y_drift <- sim_RWdrift(n, sigma, drift = 1, y0 = 0)

## -------------------------------------------------------------------------
##  y_t = 1 + 0.8 y_{t-1} + e_t  
## -------------------------------------------------------------------------

sim_AR1_stoch <- function(n, sigma = 1, phi = 0.8, c = 1, y0 = 0) {
  e <- rnorm(n, mean = 0, sd = sigma)
  y <- numeric(n)
  y[1] <- c + phi * y0 + e[1]
  for (tt in 2:n) y[tt] <- c + phi * y[tt - 1] + e[tt]
  ts(y)
}
y_vi_stoch <- sim_AR1_stoch(n, sigma, phi = 0.8, c = 1, y0 = 0)

## -------------------------------------------------------------------------
##  y_t = y_{t-1} + e_t   -> caminata aleatoria SIN deriva
## -------------------------------------------------------------------------
sim_RW <- function(n, sigma = 1, y0 = 0) {
  e <- rnorm(n, mean = 0, sd = sigma)
  y <- numeric(n)
  y[1] <- y0 + e[1]
  for (tt in 2:n) y[tt] <- y[tt - 1] + e[tt]
  ts(y)
}
y_vii <- sim_RW(n, sigma, y0 = 0)

## =========================================================================
## Gráficas de las 8 trayectorias simuladas
## =========================================================================
par(mfrow = c(3, 3), mar = c(4, 4, 3, 1))

plot(et,   main = expression(paste(y[t]==e[t], "  RB")),
     ylab = expression(y[t]), xlab = "t", col = "steelblue", type="l",lwd = 1.5)

plot(y_i,   main = expression(paste y[t]==e[t]+0.5*e[t-1], "  (MA(1))")),
     ylab = expression(y[t]), xlab = "t", col = "steelblue", lwd = 1.5)

plot(y_ii,  main = expression(y[t]==Z, ",  Z~N(0,1)")),
     ylab = expression(y[t]), xlab = "t", col = "steelblue", lwd = 1.5)

plot(y_iv,  main = expression(y[t]==sqrt(t)*e[t])),
     ylab = expression(y[t]), xlab = "t", col = "steelblue", lwd = 1.5)

plot(y_v,   main = expression(y[t]==(-1)^t*Z)),
     ylab = expression(y[t]), xlab = "t", col = "steelblue", lwd = 1.5, type = "l")

plot(y_drift, main = expression(paste(y[t]==1+y[t-1]+e[t], "  (RW con deriva)")),
     ylab = expression(y[t]), xlab = "t", col = "firebrick", lwd = 1.5)


plot(y_vii, main = expression(paste(y[t]==y[t-1]+e[t], "  (RW sin deriva)")),
     ylab = expression(y[t]), xlab = "t", col = "firebrick", lwd = 1.5)

plot(y_vi_stoch, main = "AR(1) estocástico (referencia)",
     ylab = expression(y[t]), xlab = "t", col = "darkgreen", lwd = 1.5)

par(mfrow = c(1, 1))
```
